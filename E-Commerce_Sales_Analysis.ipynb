## Core KPI

```python
query='''
SELECT
    SUM(LineValue) AS total_transaction_value,
    SUM(Quantity) AS total_units,
    COUNT(DISTINCT InvoiceNo) AS invoices,
    COUNT(DISTINCT CustomerID) AS customers,
    COUNT(DISTINCT StockCode) AS products
FROM sales;
'''
```

```python
result=con.execute(query).fetch_df()
result
```

## COUNTRY ANALYSIS

```python
Q2='''
SELECT
    Country,
    SUM(LineValue) AS transaction_value,
    SUM(Quantity) AS units,
    COUNT(DISTINCT InvoiceNo) AS invoices,
    COUNT(DISTINCT CustomerID) AS customers
FROM sales
GROUP BY Country
ORDER BY transaction_value DESC;
'''
```

```python
r2=con.execute(Q2).fetch_df()
r2
```

```python
Q3='''
SELECT
    DATE_TRUNC('month', InvoiceDate) AS month,
    SUM(LineValue) AS transaction_value,
    SUM(Quantity) AS units,
    COUNT(DISTINCT InvoiceNo) AS invoices,
    COUNT(DISTINCT CustomerID) AS customers
FROM sales
GROUP BY DATE_TRUNC('month', InvoiceDate)
ORDER BY month;
'''
```

```python
r3=con.execute(Q3).fetch_df()
r3
```

## Top Customers

```python
Q4='''
SELECT
    CustomerID,
    SUM(LineValue) AS transaction_value,
    SUM(Quantity) AS units,
    COUNT(DISTINCT InvoiceNo) AS invoices,
    COUNT(DISTINCT StockCode) AS products
FROM sales
WHERE CustomerID IS NOT NULL
GROUP BY CustomerID
ORDER BY transaction_value DESC
LIMIT 20;
'''
```

```python
r4=con.execute(Q4).fetch_df()
r4
```
